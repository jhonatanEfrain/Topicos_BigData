3. loops

```python
palabras = cadena_1.strip(',').split()
palabras
```

```python
for palabra in palabras:
    print(palabra)
```

```python
#colocar numeracion a cada palabra
for i, palabra in enumerate(palabras):
    print(f"{i+1}. {palabra}")
    
```
